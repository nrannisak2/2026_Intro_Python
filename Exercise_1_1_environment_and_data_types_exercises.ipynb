## Exercise 2: Unit conversions

A temperature is recorded as 18.5 °C. Compute it in kelvin and in degrees Fahrenheit. Separately, a discharge of 4.7 m³ s⁻¹ drains a catchment of area 250 km². Express the specific discharge in mm per day (use: 1 m³ s⁻¹ spread over 1 km² for one day equals 86.4 mm). Print each result to two decimals with an f-string.

````python
# Your solution here

temp_celcius = 18.5
temp_kelvin = temp_celcius + 273.15
temp_fahrenheit = 9 / 5 * temp_celcius + 32
print(f"temperature in K: {temp_kelvin:.2f} K")
print(f"temperature in F: {temp_fahrenheit:.2f} F")

discharge_m3_s = 4.7 #m^3/s
area = 250 #km^2

discharge_per_area = discharge_m3_s / area
print(f"discharge per area = {discharge_per_area}")

discharge_mm_day = discharge_per_area * 86.4 #mm/day
print (f"specific discharge = {discharge_mm_day:.2f} mm/day")


````

## Exercise 3: Operator families and a quality flag

Three daily readings arrive from a station: `reading_1_celsius` = 18.2, `reading_2_celsius` = -300.0, and `reading_3_celsius` = None (the sensor reported nothing).

1. Build a boolean `is_plausible` that is `True` when the first reading is above absolute zero and below 60 °C.
2. Build `is_implausible` for the second reading: apply the same test as in step 1, and negate it with `not`.
3. Use `is` to check whether the third reading is missing.
4. Starting from `n_valid = 0`, use `+=` twice to count how many of the first two readings pass the step 1 test. Note that `int()` casts `True` to 1 and `False` to 0.
5. The sensor is rated for -40 °C to 60 °C, both bounds included. Use `>=` and `<=` to test whether the first reading lies in that range, and `!=` to check that the first two readings differ.

````python
# Your solution here
reading_1_celsius = 18.2
reading_2_celsius = -300.0
reading_3_celsius = None

# Build a boolean is_plausible that is True when the first reading is above
# absolute zero and below 60 °C.
is_plausible = 0 < reading_1_celsius < 60
print(f"is_plausible: {is_plausible}")

# Build is_implausible for the second reading: apply the same test as in step 1,
# and negate it with not
is_implausible = not ( 0 < reading_2_celsius < 60 )
print (f"is_implausible: {is_implausible}")

# Use is to check whether the third reading is missing.
is_missing = reading_3_celsius is None
print (f"is_missing: {is_missing}")

# number 4
# to clarify: Note that int() casts True to 1 and False to 0.
n_valid = 0
n_valid += int(0 < reading_1_celsius < 60)
n_valid += int(0 < reading_2_celsius < 60)
print(f"n_valid: {n_valid}")

# number 5
sensor_range = -40 <= reading_1_celsius <= 60
reading_differ = reading_1_celsius != reading_2_celsius
print(f"sensor_range: {sensor_range}")
print(f"reading_differ: {reading_differ}")
````

## Exercise 4: Parse a record

## Exercise 7: Round-trip through a file, with an append

Write the Jungfraujoch week from the lecture — the daily maximum temperatures -8.4, -8.9, -9.2,

3. Read the file back with `.readlines()`, cast each of the seven lines to `float` by name, and

````python
# Your solution here
jsj_day1= -8.4
jsj_day2= -8.9
jsj_day3= -9.2
jsj_day4= -9.0
jsj_day5= -9.5

# Write the five values with mode "w", one .write() call per value.
from pathlib import Path
data_jsj = Path("_files/jsj_week.txt")
data_jsj.parent.mkdir(exist_ok=True)
print(f"does the file exist? {data_jsj.exists()}")

with data_jsj.open("w", encoding="utf-8") as fhandler:
  fhandler.write(f"{jsj_day1}\n")
  fhandler.write(f"{jsj_day2}\n")
  fhandler.write(f"{jsj_day3}\n")
  fhandler.write(f"{jsj_day4}\n")
  fhandler.write(f"{jsj_day5}\n")

print(f"does the file exist now? {data_jsj.exists()}")
with data_jsj.open("r", encoding="utf-8") as fhandler:
  raw_jsj = fhandler.read()

print(f"raw jsj data: {raw_jsj}")
print(type(raw_jsj))

# Reopen the file with mode "a" and
# append the last two days of the week, -7.8 and -6.1.

with data_jsj.open("a", encoding="utf-8") as fhandler:
  fhandler.write("-7.8\n")
  fhandler.write("-6.1\n")

# Read the file back with .readlines(),
# cast each of the seven lines to float by name,

with data_jsj.open("r", encoding="utf-8") as fhandler:
  read_jsj = fhandler.readlines()

print(f"read jsj:, {read_jsj}")

day1 = float(read_jsj[0])
day2 = float(read_jsj[1])
day3 = float(read_jsj[2])
day4 = float(read_jsj[3])
day5 = float(read_jsj[4])
day6 = float(read_jsj[5])
day7 = float(read_jsj[6])

print(f"check_float, {type(read_jsj)}")

# and compute the mean by summing the seven values by hand with +
# and dividing by len(). Print it to two decimals.

sum_jsj_data = day1 + day2 + day3 + day4 + day5 + day6 + day7
print(f"sum jsj data: {sum_jsj_data}")
num_lines = len(read_jsj)
print(f"total lines: {num_lines}")

mean_jsj = sum_jsj_data / num_lines
print(f"mean jsj: {mean_jsj:.2f}")
````
